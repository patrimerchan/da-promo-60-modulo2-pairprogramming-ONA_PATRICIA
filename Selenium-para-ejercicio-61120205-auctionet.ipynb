Ejemplo de código a través de Selenium con explicaciones detalladas de L'Éclaireur (mon Mistral)

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time
```

```python

# Configura el driver de Chrome y abre el navegador
driver = webdriver.Chrome()
url = 'https://auctionet.com/en/search/29-sculptures-bronzes?order=bid_desc'
 
driver.get(url)


# Busca el botón por el texto visible y haz clic
accept_button = driver.maximize_window()

time.sleep(3)

driver.find_element('css selector', 'body > form > div > div.modal.test-cookie-consent-screen-1 > div.modal__footer.modal__footer--with-actions > button:nth-child(2)').click()

print("Botón de cookies aceptado.")


# Guarda la altura inicial de la página
last_height = driver.execute_script("return document.body.scrollHeight")

# Bucle para simular scroll hasta el final de la página
while True:
    # Scroll hasta el final
    driver.find_element(By.TAG_NAME, 'body').send_keys(Keys.END)
    # Espera a que se cargue el contenido
    time.sleep(2)
    # Calcula la nueva altura y compara con la anterior
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height:
        break
    last_height = new_height

# Extrae todos los títulos (incluyendo los nuevos después del clic)
titulos = driver.find_elements('css selector', "span.test-item-title")
for titulo in titulos:
    print("Texto visible:", titulo.text)
    print("Valor del atributo title:", titulo.get_attribute("title"))
    print("---")  # Separador para mayor claridad


# Cierra el navegador
driver.quit()

```

Explicación de los comentarios

Importaciones:
Cada librería tiene un propósito específico, desde controlar el navegador hasta simular teclas.

Configuración del driver:
Abre Chrome y navega a la URL deseada.

Espera inicial:
Da tiempo a que la página se cargue antes de interactuar con ella.

Scroll infinito:
Simula el scroll hasta el final de la página y espera a que se cargue nuevo contenido.

Extracción de títulos:
Busca y muestra todos los títulos de los artículos.

Cierre del navegador:
Libera los recursos al finalizar.

Si queréis adaptar este código para extraer otros datos (como bids, días restantes o precios), solo tenéis que cambiar el selector en find_elements y ajustar la lógica según la estructura de la página.
